```python
from google.colab import files
import pandas as pd
uploaded=files.upload()
df=pd.read_csv('Sales_data.csv')
df.head()
```

```python
df.rename(columns={
    'Order_ID':'OrderID'
}, inplace=True)

print("\n AFTER REMAINING")
print(df)
```

```python
df.dropna(subset=['Quantity'], inplace=True)
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
df
```

```python
print("\n CLEANED DATASET")
print(df)

print("\n FINAL SHAPE:",df.shape)
```

```python
print(df.isnull().sum())
```

```python
df.dropna()
```

```python

df["Amount"]=df["Amount"].fillna(df["Amount"].median())
print(df)
```

```python
df.duplicated()
```

```python
import numpy as np
filtered=df[df["Amount"]>10000]
print("/nFiltered Data(Amount>10000)")
print(filtered)
```

```python
sorted_df=df.sort_values(by="Amount",ascending=False)
print("\nSorted Data(Amount in Descending Order)")
print(sorted_df)
```

```python
grouped=df.groupby("Product")["Amount"].sum()
print("/nGroupby product")
print(grouped)
```

```python
agg=df["Amount"].agg(["sum","mean","min","max"])
print("\nAggregation")
print(agg)
```

```python
region_sales=df.groupby("Region")["Amount"].sum()
print("\nRegion wise Amount")
print(region_sales)
```

```python

top_products = df.groupby('Product')['Amount'].sum().sort_values(ascending=False)

print(top_products)
```

```python
monthly_sales = df.groupby('Month')['Amount'].sum()
print(monthly_sales)
```

```python
import matplotlib.pyplot as plt
top_products.plot(kind='bar')
plt.title('Top Selling Products')
plt.xlabel('Product')
plt.ylabel(' Sales Amount')

plt.show()
```

```python
monthly_sales.plot(kind='line',marker='o')

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales Amount')

plt.show()
```
